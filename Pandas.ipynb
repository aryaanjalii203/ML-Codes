```python
data={
    "name":["aman","riya","gia","david "],
    "age":[20,21,23,19],
    "marks":[80,89,92,78]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(df.loc[[1,3]])
```

```python
print(df.iloc[[1,3]])
```

```python
higher_score=df[df["marks"]>85]
print(higher_score)
```

```python
df["passed"]=df["marks"]>40
print(df)
```

```python
df["Extra_grades"]=df["marks"]+5
print(df)
```

```python
data={
    "name":["Tanya","Rajiv","Ravi"],
    "marks":[85,None,92]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
df['marks']=df["marks"].fillna(df["marks"].mean())
print(df)
```

```python
print(df.dropna())
```

```python
print(df.sort_values("marks",ascending=False))
```

```python
print(df["marks"].max())
```

```python
print(df["marks"].min())
```

```python
print(df['marks'].mean())
```
